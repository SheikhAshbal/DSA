# Execution order and reproducibility

Run all cells from a fresh kernel. The five-row fixture is synthetic. The historical A–C–B–D run would print (5, 2) in C; the clean run below prints (3, 2) in both C and D.

```python
import pandas as pd

def average(values):
    return sum(values) / len(values)

print(average([2, 4, 6]))
try:
    average([2, 4, '6'])
except TypeError:
    print('TypeError: mixed numeric and string input')
```
