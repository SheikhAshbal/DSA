## Understanding the Dataset

It is **important** to inspect the data before starting the *analysis*.

| Variable | Type | Description |
| --- | --- | --- |
| age | Integer | Age of the customer in years |
| spend | Float | Amount spent by the customer |

I use the Python function `print()` to display results, and I can find more information on the [Python website](https://www.python.org/).

We standardise a feature using the following formula:

$$
z = \frac{x - \mu}{\sigma}
$$

Here, $x$ is the original value, $\mu$ is the mean, and $\sigma$ is the standard deviation. For machine learning, the mean and standard deviation are calculated from the training data only.
